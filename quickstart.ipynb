### Step 2: Ingest Live 24/7 On-Chain Base Money (M)
Queries DefiLlama aggregate stablecoins (USDT + USDC + DAI) to determine systemic liquidity expansion or contraction.

```python
import sys
sys.path.insert(0, '.')
from core.onchain_liquidity import OnChainLiquidityEngine
import datetime, time
import matplotlib.pyplot as plt

liq_engine = OnChainLiquidityEngine()
dates = [(datetime.datetime.now(datetime.timezone.utc) - datetime.timedelta(days=i)).strftime('%Y-%m-%d') for i in range(180, -1, -1)]
features = liq_engine.compute_rolling_features(dates, window=30)

dates_plot = [f['date'] for f in features]
z_scores = [f['z_score'] for f in features]

plt.figure(figsize=(10, 4), dpi=100)
plt.plot(dates_plot[::7], z_scores[::7], color='#2563eb', lw=2, label='Stablecoin Float Z-Score (252-Day Window)')
plt.axhline(0.8, color='#10b981', linestyle='--', label='Decoupling Expansion (+0.8 sigma)')
plt.axhline(-1.5, color='#ef4444', linestyle='--', label='Liquidity Crunch Threshold (-1.5 sigma)')
plt.title('Systemic 24/7 On-Chain Base Money Float (M)', fontsize=12, fontweight='bold')
plt.ylabel('Standard Deviations (Z)')
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

latest = features[-1]
print(f"Latest Date: {latest['date']} | Z-Score: {latest['z_score']:.2f} sigma | 30D Float Growth: {latest['velocity_30d']*100:+.2f}%")
```
